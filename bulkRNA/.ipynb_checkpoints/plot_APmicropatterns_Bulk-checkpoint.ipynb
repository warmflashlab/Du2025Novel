# Plots related to bulk RNA seq dataset

Dataset was generated and analyzed by Siqi Du

Cell line: ESI017

Condition: during neural differentiation (N2B27, SB, LDN), WNT3A is added at 34h, samples are collected at timepoints 34h, 42h, 50h, 72h

```python
# reload .csv file of the combined
import numpy as np
import pandas as pd
import math
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib.font_manager as fm
import matplotlib.colors as mcolors
from pathlib import Path
base_dir = Path("/Users/dududu/Library/CloudStorage/Box-Box/Siqi/bulkRNAseq")
data = pd.read_csv(base_dir / "AP_neural_Wnt_FGF/data/abundance_combined_W.csv")
data.drop(data.columns[0], axis=1, inplace=True)
```

```python
# function will be used in this analysis
def get_value(df_exp, gene_name, cond_name):
    gene_idx = df_exp.index[df_exp["Gene name"] == gene_name].tolist()
    return np.array(df_exp.iloc[gene_idx][cond_name])[0]

def get_gene_exp_mat(df_exp, gene_name, time_points=[34,42,50,72]):
    df_repl = []
    for tp in time_points: 
        df_repl.append({'Time': tp,
                        'R1': get_value(df_exp, gene_name, f"H{tp}_R1"),
                        'R2': get_value(df_exp, gene_name, f"H{tp}_R2")})
    df_repl = pd.DataFrame(df_repl)
    return df_repl

def plot_genes(df_exp, gene_names, title_name,
               colors=sns.color_palette("muted"), errorbar_display=True, legend_display=True, show=True):
    
    fig, ax = plt.subplots(figsize=(5,4.5))
    ax.set_xlabel("Time (h)", fontsize=20)
    ax.set_ylabel("Expression (TPM)", fontsize=20)

    time_points=[34,42,50,72]
    for idx,gene in enumerate(gene_names):
        df_repl = get_gene_exp_mat(df_exp, gene_names[idx])
        array_mean = df_repl[["R1","R2"]].mean(axis=1).to_numpy()
        array_se = df_repl[["R1","R2"]].std(axis=1).to_numpy()/math.sqrt(2)            
        if errorbar_display:
            plt.errorbar(time_points, array_mean, yerr=array_se, color=colors[idx], linewidth=2)
        else:
            plt.plot(time_points, array_mean, color=colors[idx], linewidth=1.5)
    
    font_props = fm.FontProperties(style='italic', size=18)
    if legend_display:
        plt.legend(gene_names, prop=font_props) 
    fig.savefig(base_dir / f"APmicropatterns/LinePlot-{title_name}.png", dpi=300, bbox_inches="tight")
    if not show:
        plt.close(fig)

def plot_gene_tiny(df_exp, gene_name, title_name, top_space_scale=0,
                   colors=sns.color_palette("muted"), show=True, xtickshow=True):
    
    fig, ax = plt.subplots(figsize=(2.4,2.5))
    time_points=[34,42,50,72]
    df_repl = get_gene_exp_mat(df_exp, gene_name)
    array_mean = df_repl[["R1","R2"]].mean(axis=1).to_numpy()
    array_se = df_repl[["R1","R2"]].std(axis=1).to_numpy()/math.sqrt(2)
    plt.errorbar(time_points, array_mean, yerr=array_se, color=colors, linewidth=2)
    ymin, ymax = ax.get_ylim()
    ax.set_ylim(ymin, ymax+ (ymax-ymin) * top_space_scale)
    if not xtickshow:
        ax.set_xticks([])
    ax.text(0.98, 0.98,gene_name,transform=ax.transAxes,ha="right",va="top",fontsize=16,fontstyle="italic")
    fig.savefig(base_dir / f"APmicropatterns/LinePlot-{title_name}.png", dpi=300, bbox_inches="tight")
    if not show:
        plt.close(fig)    
```

## Plot the change of gene expression

```python
# swarm or boxen plot - the change of gene expression between time points
# step 1: select a subset of genes and filter out unexpressed ones
# step 2: get the change of expression within replicate between adjacent timepoints (e.g., delta = H_T2_R1 - H_T1_R1)
# step 3: prepare a dataframe with mean values in required format and label
# step 4: plot
```

```python
# step 1
selected_genes = ["MYC","CCND1","TCF7","LEF1","PPARD","JUN","FOSL1","MMP7","AXIN2","NRCAM","TCF4","CLDN1","VEGFA","FGF1","MYCBP",
                  "ID2","TERT","LGR5","FZD7","FST","FN1","EN2","GJA1","TBX1","TBX3","NEUROG1","SP5","NEUROD1","GBX2","CDX1",
                  "CDC25A","PITX2","KRT1","CTLA4","FGF4","VCAN","TNFRSF19"] + ["WNT8A"] + ["HOXA1","HOXA2","HOXA3","HOXA4","HOXA5",
                  "HOXA6","HOXA7","HOXB1","HOXB2","HOXB3","HOXB4","HOXB5","HOXB6","HOXB7","HOXB8","HOXC4","HOXC5","HOXC6","HOXC8",
                  "HOXD1","HOXD3","HOXD4","HOXD8"]
data_aux = data[data['Gene name'].isin(selected_genes)].copy()

### filter out unexpressed genes
### standard for expressed genes: there is one condition (time point) that abundance (TPM) >= 1 in both replicates
time_points = [34,42,50,72]
idx_exp = False
for tp in time_points:
    both_high = (data_aux[f'H{tp}_R1'] >= 1) & (data_aux[f'H{tp}_R2'] >= 1)
    idx_exp |= both_high   # OR logic

data_filtered = data_aux[idx_exp]
```

```python
# step 2
data_filtered['D42-34_R1'] = data_filtered['H42_R1'] - data_filtered['H34_R1']
data_filtered['D42-34_R2'] = data_filtered['H42_R2'] - data_filtered['H34_R2']
data_filtered['D50-42_R1'] = data_filtered['H50_R1'] - data_filtered['H42_R1']
data_filtered['D50-42_R2'] = data_filtered['H50_R2'] - data_filtered['H42_R2']
data_filtered['D72-50_R1'] = data_filtered['H72_R1'] - data_filtered['H50_R1']
data_filtered['D72-50_R2'] = data_filtered['H72_R2'] - data_filtered['H50_R2']
```

```python
# step 3
# note: in order to prepare for swarm/box/boxen plot, the dataframe columns are:
# gene name, time interval, delta replicate 1, delta replicate 2, gene class (Wnt target, WNT8A, HOX), delta mean, standard error
time_intervals = ['H34-42','H42-50','H50-72']
data_aux = data_filtered.copy()

### transform
df1 = pd.DataFrame({
    'Gene name': data_aux.iloc[:, 1].values.repeat(3),
    'Time interval': time_intervals * len(data_aux),
    'Delta R1': data_aux.iloc[:, [10,12,14]].to_numpy().ravel(),
    'Delta R2': data_aux.iloc[:, [11,13,15]].to_numpy().ravel()
})

### catogorize genes
category_1 = ["MYC","CCND1","TCF7","LEF1","PPARD","JUN","FOSL1","MMP7","AXIN2","NRCAM","TCF4","CLDN1","VEGFA","FGF1","MYCBP","ID2",
              "TERT","LGR5","FZD7","FST","FN1","EN2","GJA1","TBX1","TBX3","NEUROG1","SP5","NEUROD1","GBX2","CDX1","CDC25A","PITX2",
              "KRT1","CTLA4","FGF4","VCAN","TNFRSF19"]
category_2 = ["WNT8A"]
category_3 = ["HOXA1","HOXA2","HOXA3","HOXA4","HOXA5","HOXA6","HOXA7","HOXB1","HOXB2","HOXB3","HOXB4","HOXB5","HOXB6","HOXB7",
              "HOXB8","HOXC4","HOXC5","HOXC6","HOXC8","HOXD1","HOXD3","HOXD4","HOXD8"]

df1['Gene class'] = 'Unclassified'
df1.loc[df1.iloc[:, 0].isin(category_1), 'Gene class'] = 'Wnt target'
df1.loc[df1.iloc[:, 0].isin(category_2), 'Gene class'] = 'WNT8A'
df1.loc[df1.iloc[:, 0].isin(category_3), 'Gene class'] = 'HOX'

### calculate mean
df1['Delta mean'] = (df1['Delta R1'] + df1['Delta R2']) / 2

### filter out genes has no change (delta = 0)
df1 = df1.groupby('Gene name').filter(
    lambda g: not (g['Delta mean'] == 0).all())
```

```python
# step 4
plt.figure(figsize=(5, 4))
mypalette = [[0.9,0.65,0.5],[0.7, 0, 0],[0.7, 0.7, 0.7]]
sns.swarmplot(df1, x="Time interval", y="Delta mean", size=6, 
              hue="Gene class", hue_order=['Wnt target','WNT8A','HOX'],
              palette = mypalette)
plt.savefig(base_dir / "APmicropatterns/SwarmPlot-expression-in-categories.png", dpi=300, bbox_inches='tight')
```

```python
ax = sns.boxplot(data=df1, x="Time interval", y="Delta mean", 
            hue="Gene class", hue_order=['Wnt target','WNT8A','HOX'],
            palette=mypalette, fill=False, gap=.1, linewidth=1.5)

ax.axvline(0.5, color=".7", dashes=(2, 2), linewidth=1)
ax.axvline(1.5, color=".7", dashes=(2, 2), linewidth=1)

plt.yscale('symlog')
ax.margins(y=0.075)
plt.xlabel("Time interval", fontsize=20)
plt.ylabel("\u0394 expression", fontsize=20)
plt.xticks(fontsize=18)
plt.yticks(fontsize=14)
leg = plt.legend(title="Gene class", fontsize=18, title_fontsize=20, bbox_to_anchor=(1.05, 0.5), loc='center left', frameon=False)
for text in leg.get_texts():
    text.set_fontstyle("italic")

plt.savefig(base_dir / "APmicropatterns/BoxPlot-expression-in-categories.png", dpi=300, bbox_inches='tight')
```
